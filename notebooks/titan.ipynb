```python
# Import the data

import pandas as pd
from pathlib import Path

path=Path('../data/train.csv')

train_df= pd.read_csv(path)

path2=Path('../data/test.csv')

test_df= pd.read_csv(path2)

# Get a glimpse of the data and its vital metrics 
def preprocess(data):
    print('='*30,'Head','='*30)
    display(data.head(10))

    print('='*30,'Info','='*30)
    print(data.info())

    print('='*30,'Numeric summary stats','='*30)
    display(data.describe(include=['int','float']))
    
    print('='*30,'String summary stats','='*30)
    display(data.describe(include=['str','object'])) #Once the data is encoded this will generate an  error but just ignore it

    return

preprocess(train_df)
```

```python
#Survival rate by class
train_df.groupby('Pclass')['Survived'].mean()
```

```python
#Survival rate by sex
train_df.groupby('Sex')['Survived'].mean()
```

```python
#Survival rate by sex
#train_df.groupby('SibSp')['Survived'].mean()

train_df['family_size']= 1 + train_df['SibSp']
family_map=  {1:'Alone',2:'Small',3:'Small',4:'Small',5:'Medium',6:'Medium',7:'Medium',8:'Large',9:'Large',10:'Large',11:'Large'}

train_df['fam_size'] = train_df['family_size'].map(family_map)
```

```python
# Survival rate by fam size
train_df.groupby('fam_size')['Survived'].mean()
```

```python
def datacleaner(dataframe):
    dataframe = dataframe.copy()

    # Drop unnecessary columns
    dataframe.drop(['Name', 'Ticket', 'Cabin'], axis=1, inplace=True)

    # Fill missing numerical values
    dataframe['Age'] = dataframe['Age'].fillna(dataframe['Age'].median())

    # Fill missing categorical values
    dataframe['Embarked'] = dataframe['Embarked'].fillna(
        dataframe['Embarked'].mode()[0]
    )

    # Encode Sex
    dataframe['Sex'] = dataframe['Sex'].map({
        'male': 0,
        'female': 1
    })

    # Encode Embarked
    dataframe = pd.get_dummies(
        dataframe,
        columns=['Embarked'],
        drop_first=True,
        dtype=int
    )

    return dataframe
```

```python
preprocess(test_df)
```

```python
#Clean train_df
train_df=datacleaner(train_df)

#Clean test_df
test_df=datacleaner(test_df)
```

```python
#Look into the dataframes to confirm uniformity
preprocess(train_df)
preprocess(test_df)
```

```python
#Spliting our data into training and testing

target_column = "Survived"

X_train = train_df.drop(columns=[target_column]) #<---Train data from train_csv

y_train = train_df[target_column] #<---Train data from train_csv

X_test = test_df.copy() #<-- Test data from test_csv
```

```python
#Spliting out train data for evaluation
from sklearn.model_selection import train_test_split

X_tr, X_val, y_tr, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42,stratify=y_train)
```

```python
#Training a dummy model

from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import make_pipeline
from sklearn.dummy import DummyClassifier

baseline_mode=make_pipeline(
    DummyClassifier(strategy='most_frequent')
)

baseline_mode.fit(X_tr,y_tr)

y_pr= baseline_mode.predict(X_val)

display(baseline_mode.coef_)
```

```python
#Evaluate the baseline model
import matplotlib.pyplot as plt
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

# A function for model evaluation
def evaluate_model(y_test,y_pred,model=None,X_train=None,y_train=None,X_test=None,):


    """
    Evaluate a classification model using multiple performance metrics.

    Parameters
    ----------
    y_val : array-like
        Actual target values from the test dataset.

    y_pred : array-like
        Target values predicted by the classification model.

    model : estimator, optional
        Trained machine learning model. Used to calculate training
        and testing scores.

    X_val : array-like, optional
        Test features used to calculate the model's test score.

    X_tr : array-like, optional
        Training features used to calculate the model's training score.

    Returns
    -------
    None
        Prints model scores, accuracy, precision, recall, F1-score,
        confusion matrix, confusion matrix display, and classification report.
    """


    print('='*30,'Training score','='*30)
    print(model.score(X_train,y_train))
    print('='*30,'Test score','='*30)
    print(model.score(X_test,y_test))
    print('='*30,'Accuracy score','='*30)
    print(accuracy_score(y_test,y_pred))
    print('='*30,'Precision Score','='*30)
    print(precision_score(y_test,y_pred))
    print('='*30,'Recall score','='*30)
    print(recall_score(y_test,y_pred))
    print('='*30,'f1_score','='*30)
    print(f1_score(y_test,y_pred))
    print('='*30,'Confusion_matrix','='*30)
    print(confusion_matrix(y_test,y_pred))
    print('='*30,'Confusion_matrix_Display','='*30)
    cm=ConfusionMatrixDisplay(confusion_matrix=confusion_matrix(y_test,y_pred))
    print(cm)
    plt.show(cm)
    print('='*30,'Classififcation Report','='*30)
    print(classification_report(y_test,y_pred))

evaluate_model(y_val,y_pr,baseline_mode,X_tr,y_tr,X_val)
```

```python
# Using cross validation to find a model that fits best
from sklearn.model_selection import StratifiedKFold,cross_validate,cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier


models = {
    'Logistic Regression': LogisticRegression(max_iter=1000),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42),
    'SVM': SVC(),
    'KNN': KNeighborsClassifier()
}

cv= StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=13
)

results=[]

criteria= ['accuracy','precision','recall','f1']

for name,model in models.items():

    scores=cross_validate(
        model,
        X_tr,
        y_tr,
        cv=cv,
        scoring=criteria
    )

    results.append({
        'Model': name,
        'Accuracy': scores['test_accuracy'].mean(),
        'Accuracy_std': scores['test_accuracy'].std(),
        'Precision': scores['test_precision'].mean(),
        'Precision_std': scores['test_precision'].std(),
        'Recall': scores['test_recall'].mean(),
        'Recall_std': scores['test_recall'].std(),
        'F1': scores['test_f1'].mean(),
        'F1_std': scores['test_f1'].std()
    })

results_df = pd.DataFrame(results)
```

```python
# Hyperparameter tuning using Random search (this is computational heavy and takes some time)
from sklearn.model_selection import GridSearchCV,RandomizedSearchCV

model= RandomForestClassifier(random_state=42)

param_distributions = {
    'n_estimators': [20,50, 100, 150, 200],
    'max_depth': [1,2,3,4],
    'min_samples_split': [5, 10, 15, 20],
    'min_samples_leaf': [2, 4, 6, 8],
    'max_features': ['sqrt', 'log2',]
}

grid_search= RandomizedSearchCV(
    estimator=model,
    n_iter=60,
    param_distributions=param_distributions,
    n_jobs=-1,
    verbose=1,
    scoring='f1',
    cv=cv
)

fitted_grid=grid_search.fit(X_tr,y_tr)
```

```python
print(f'The best score is {grid_search.best_score_}')
print(f'The best params are {fitted_grid.best_params_}')
print(f'The best estimator is {fitted_grid.best_estimator_}')
```

```python
# Operating on the final model

from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier

#Instance a model( hyperparams derived from cells bellow)
proper_model= RandomForestClassifier( max_depth=4, min_samples_leaf=2, min_samples_split=20,
                        n_estimators=150, random_state=42)

#Scale the data
scaler= StandardScaler()
X_train_scaled= scaler.fit_transform(X_train)
X_test_scaled= scaler.transform(X_test) #transform only

#Fit the model
proper_model.fit(X_train_scaled,y_train)
```

```python
#Generate predictions
Survived = proper_model.predict(X_test_scaled)

# Create a submission/results dataframe
submission_df = pd.DataFrame({
    'PassengerId': test_df['PassengerId'], 
    target_column: Survived
})

# Save to a new CSV file
submission_df.to_csv("bethuel_predictions.csv", index=False)
print("Predictions saved successfully!")
```
